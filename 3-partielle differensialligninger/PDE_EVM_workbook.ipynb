## 1D Diffusjonsligning og EVM

I forelesning har vi sett på en rekke av eksempler i 1D for å fremstille endelig volum metode (EVM). Vi har startet med en veldig enkelt 1D diffusjonsligning

$$\frac{d}{dx}\left(k \frac{d}{dx}\phi \right) + S = 0$$

Det første eksempel vi har sett på var uten kildeterm, dvs. $S=0$ og $k$ konstant. Koden som implementerer EVM på en ikke uniform inndeling ligger i neste cellen:

```python
m = 5
x = np.random.rand(m+2)*0.5
x = np.sort(x)
x[0]  = 0
x[-1] = 0.5
#x = np.array([0,0.05,0.15,0.25,0.35,0.45,0.5]) # eksempel fra Versteeg
kA = 10      # termisk diffusitivtet * areal på tverrsnitt
dx = x[1 : ] - x[ : -1]
print(dx)
a = 100
b = 500

L = kA*(-np.diag((1/dx)[1:-1],  1) - np.diag(1/dx[1:-1], -1) + np.diag(1/dx[1:], 0) + np.diag(1/dx[:-1], 0)) 
F = np.zeros((m))
F[0]  = kA*1/dx[0]*a
F[-1] = kA*1/dx[-1]*b

u = la.solve(L, F)

plt.plot(x[1:-1],u)
```

Etter første eksemplet hat vi sett på en eksempel (3.12 i forelesningsnotater) av en plate av bredde $L=2$cm, med oppvarming som skjer i platen, hvor styrende ligning er 

$$\frac{d}{dx}\left(k \frac{d}{dx}\phi \right) + q = 0, \quad T_V=100^\circ C, \quad T_H = 200^\circ C$$

Problemet medførte de konstanter
$$ k = 0.5, A = 1, q = 1000 kW/m3 = 10^6 W/m3, \delta x = 0.004 m$$

Så vi kan beregne $kA/\delta x$ osv. og får ligningssystemet i matriseform

$$
\begin{bmatrix}
375 & -125 & 0 & 0&0 \\
-125 & 250 & -125 & 0 & 0 \\
0& -125& 250 & -125 & 0\\
0&0&-125& 250 &-125 \\
0&0&0&-125&375
\end{bmatrix} \begin{bmatrix} T_1 \\ T_2 \\ T_3 \\ T_4 \\ T_5 \end{bmatrix} = \begin{bmatrix} 
29000\\ 4000 \\ 4000 \\ 4000 \\ 54000\end{bmatrix}
$$

### Husk: Viktige ligninger

$$a_P = a_W +a_E - S_P$$

Hvis noden er ikke en randnode så gjelder $a_W=\frac{kA}{\delta x} = a_E$. For randnoder ligningen må modifiseres (sjekk forelesningsnotater eller slides).
